### 交付物一：项目环境配置文档

#### 1. 开发环境

- 操作系统：macOS
- 环境管理工具：Anaconda
- Conda环境名称：credit-default
- Python版本：Python 3.12
- 开发工具：JupyterLab

#### 2. 使用的主要Python库

- NumPy：用于数值计算
- Pandas：用于读取和处理数据
- Matplotlib：用于绘制基础图表
- Seaborn：用于数据可视化
- SciPy：用于统计分析
- Scikit-learn：用于后续的数据处理和模型训练

#### 3. 环境创建过程

首先使用Anaconda创建名为credit-default的独立环境，然后安装数据分析需要的Python库和JupyterLab。最后将该环境注册为Jupyter Notebook内核。
